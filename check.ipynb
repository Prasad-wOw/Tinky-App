```python
def solve_n_queens(n=8):
    def backtrack(r):
        if r == n:
            solutions.append(["." * c + "Q" + "." * (n - c - 1) for c in queens])
            return
        for c in range(n):
            if c not in queens and r - c not in diag1 and r + c not in diag2:
                queens.append(c)
                diag1.add(r - c)
                diag2.add(r + c)
                backtrack(r + 1)
                queens.pop()
                diag1.remove(r - c)
                diag2.remove(r + c)

    solutions = []
    queens = []
    diag1 = set()
    diag2 = set()
    backtrack(0)
    return solutions

for sol in solve_n_queens():
    for row in sol:
        print(row)
    print()
    break

```
